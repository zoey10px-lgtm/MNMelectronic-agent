# M&M 증강 자료 164장 재학습 및 평가
받은 증강 학습 사진 164장과 기존 검증 원본 4장을 사용합니다. 이전 모델에서 이어 학습하며 위치와 유형이 맞는지 같은 기준으로 비교합니다. Colab 런타임에서 GPU를 선택하면 빠르게 학습할 수 있습니다.

```python
from google.colab import files
import pathlib, zipfile, subprocess, sys
uploaded=files.upload()
archive=next(n for n in uploaded if n.lower().endswith(".zip"))
with zipfile.ZipFile(archive) as z: z.extractall("/content/mm_repaired")
root=next(pathlib.Path("/content/mm_repaired").rglob("train_4class.py")).parent
subprocess.run([sys.executable,"-m","pip","install","-r",str(root/"requirements.txt")],check=True)
```

```python
subprocess.run([sys.executable,str(root/"train_4class.py"),"--epochs","60","--batch","8"],check=True,cwd=root)
```

```python
import json,shutil
report=json.loads((root/"MODEL_EVALUATION.json").read_text())
print(json.dumps(report["summary"],ensure_ascii=False,indent=2))
output=shutil.make_archive("/content/MM_Electronics_Evaluated","zip",root.parent,root.name)
files.download(output)
```
